```python
import matplotlib.pyplot as plt
sales_by_category=df.groupby("Category")["Sales"].sum()
sales_by_category.plot(kind="pie",autopct="%1.1f%%")
plt.ylabel("")
plt.title("Sales by Category")
plt.show()

```

```python
df.groupby("Date")["Sales"].sum()
```

```python
import matplotlib.pyplot as plt
date_sales=df.groupby("Date")["Sales"].sum()
date_sales.plot(kind="line",marker="o")
plt.xlabel("Date")
plt.ylabel("Total Sales")
plt.title("Sales over time")
plt.show()
```

```python
import matplotlib.pyplot as plt
sales_by_product=df.groupby("Product_Name")["Sales"].sum().sort_values(ascending=True).head(3)
sales_by_product.plot(kind="barh")
plt.ylabel("Product")
plt.xlabel("Total Sales")
plt.title("Top 3 products by Sales")
plt.show()
```

```python
import matplotlib.pyplot as plt
sales_by_city=df.groupby("Customer_City")["Sales"].sum()
sales_by_city.plot(kind="bar")
plt.xlabel("City")
plt.ylabel("Total Sales")
plt.title("Sales by City")
plt.show()
```

```python
import matplotlib.pyplot as plt
season_sales=df.groupby("Season")["Sales"].sum()
season_sales.plot(kind="bar")
plt.xlabel("Season")
plt.ylabel("Total Sales")
plt.title("Season_sales")
plt.show()
```

## Business Insights
1. Women Wear contributes the highest sales (around 36%).
2. Chaniya Choli is the top-selling product by revenue.
3. Total shop revenue is RS 1,28,080.
4. Sales fluctuate over time rather than showing a steady trend.
5. The highest sales were recorded on 15/09.
6. Sales increase during festive seasons like Navratri and Wedding.
7. Business can improve revenue by stocking popular products and planning offers during low-sales periods.
